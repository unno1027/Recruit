## ミニバッチ取得の改良

本講義では学習時のミニバッチ取得に `np.random.choice` を用いたが、これはエポックの定義に照らすと良い実装とは言えない

   - 利用した生成AIに関する情報

   - 生成AIに渡したプロンプト全文
   - 生成AIが出力した回答内容全文

**問1 回答欄**

np.random.choiceを用いた実装が良い実装と言えないのは、復元抽出（重複を許すサンプリング）になってしまうためである。1エポック内で同じデータが何度も選ばれる一方で一度も選ばれないデータが生じ、データセット全体を均等かつ網羅的に学習するというエポックの定義や本来の勾配降下法の挙動から外れてしまうからである。

**問2 回答欄**

ミニバッチ取得の方法として望ましい実装は、訓練データのインデックスを事前にシャッフルし、重複や偏りのないように順番に取り出す非復元抽出（無作為非復元抽出）である。データセット全体をくまなく1周（1エポック）できるようにミニバッチ単位で分割して網羅的に学習させるのが望ましい。

def make_batch(data_x, data_t, batch_size):
    data_size = data_x.shape[0] # 学習データの数
    batch_mask = np.random.choice(data_size, batch_size)
    batch_x = data_x[batch_mask]
    batch_t = data_t[batch_mask]
    return batch_x, batch_t
    
前提：「本講義では学習時のミニバッチ取得に np.random.choice を用いたが、これはエポックの定義に照らすと良い実装とは言えない」なぜ np.random.choice を用いた実装が良い実装と言えないか、100字程度で論じなさいミニバッチ取得の方法として望ましい実装はどのような実装か、100字程度で論じなさい生成AIが出力した回答内容全文（上記 問1および問2の解答内容がこれに該当します）ミニバッチ学習の理論に照らしたときの修正すべき点と根拠修正すべき点: np.random.choice(data_size, batch_size) では引数の replace が指定されていないためデフォルトで True（復元抽出）となり、同一バッチ内または同一エポック内で重複したインデックスが選ばれてしまう点。   根拠: ミニバッチ学習の1エポックは「すべての訓練データを漏れなく1回ずつ学習すること」を意味する。復元抽出では偏りが生じ、理論的に正しい確率的勾配降下法の挙動やエポックの概念を満たさなくなるため、np.arange(data_size) を事前に np.random.permutation() 等でシャッフルし、順番にスライスして取得する非復元抽出の仕組みに修正すべきである。
