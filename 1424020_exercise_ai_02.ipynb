## NumPyを用いた行列計算

1. $ A = \begin{bmatrix} 5 & 1 \\ 2 & 0 \end{bmatrix}, \space B = \begin{bmatrix} -1 & 4 \\ -4 & -9 \end{bmatrix} $ のとき、$ A -3X = B $ となる行列 $ X $ を求めるプログラムを実装せよ
2. $ A = \begin{bmatrix} 2 & 1 & 1 \\ 1 & 1 & 2 \\ 1 & 1 & 1 \end{bmatrix} $ のとき、$ A $ の逆行列を求めるプログラムを実装せよ
3. $ A = \begin{bmatrix} 6 & -3 & -7 \\ -1 & 2 & 1 \\ 5 & -3 & -6 \end{bmatrix} $ のとき、$ A $ の固有値と固有ベクトルを求めるプログラムを実装せよ

**利用するモジュール・メソッドのインポート**

```python
import numpy as np
```

```python
# 問1 回答欄
import numpy as np

# 行列 A と B の定義
A = np.array([[5, 1], [2, 0]])
B = np.array([[-1, 4], [-4, -9]])

# A - 3X = B より X = (A - B) / 3
X = (A - B) / 3

print("1. 行列 X:")
print(X)
```

```python
# 問2 回答欄
import numpy as np

# 行列 A の定義
A = np.array([[2, 1, 1], 
              [1, 1, 2], 
              [1, 1, 1]])

# 逆行列の計算
A_gyaku = np.linalg.inv(A)

print("2. 逆行列 A^(-1):")
print(A_gyaku)
```

```python
# 問3 回答欄
import numpy as np

# 行列 A の定義
A = np.array([[2, 1, 1], 
              [1, 1, 2], 
              [1, 1, 1]])

# 逆行列の計算
A_gyaku = np.linalg.inv(A)

print("2. 逆行列 A^(-1):")
print(A_gyaku)
```
